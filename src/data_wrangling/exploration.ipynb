```python
data_antidepressant = collate_all_antidepressant_data()
data_all_drugs = all_drugs()
    
```

```python
antidepressant_overview = pd.DataFrame(data_antidepressant.groupby(by='bnf_chemical_substance')['TOTAL_by_item'].sum())
antidepressant_overview = antidepressant_overview.reset_index()
antidepressant_overview.columns = ['bnf_chemical_substance','total']
antidepressant_overview = antidepressant_overview.sort_values(by='total',ascending=False)
antidepressant_overview
```

```python
antidepressant_month = data_antidepressant.pivot_table(columns = 'year_month', 
                 index = 'bnf_chemical_substance',
                 values = 'TOTAL_by_item',
                 aggfunc = 'sum',
                 fill_value = 0,
                 margins = True)
antidepressant_month
```
