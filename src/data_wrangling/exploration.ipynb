```python
import pandas as pd
import os
from utils import collate_all_antidepressant_data
# change directory to the main dir so the function could work like in other python file
%cd ../../
```

```python
data = collate_all_antidepressant_data()
data.head()
```

```python
drugs_overview = pd.DataFrame(data.groupby(by='bnf_chemical_substance')['TOTAL_by_item'].sum())
drugs_overview = drugs_overview.reset_index()
drugs_overview.columns = ['bnf_chemical_substance','total']
drugs_overview = drugs_overview.sort_values(by='total',ascending=False)
drugs_overview
```

```python
drugs_month = data.pivot_table(columns = 'yearmonth', 
                 index = 'bnf_chemical_substance',
                 values = 'TOTAL_by_item',
                 aggfunc = 'sum',
                 fill_value = 0,
                 margins = True)
drugs_month
```
